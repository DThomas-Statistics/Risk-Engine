```python
def kurtosis(percent_vec):
  mean=np.mean(percent_vec)
  avg_fourth = np.mean((percent_vec-mean)**4)
  std = np.std(percent_vec)**4
  kurtosis = avg_fourth/std
  return kurtosis-3

def a(percent_vec):
  squared_values = percent_vec**2
  k = acf(squared_values, adjusted=False, nlags=18, qstat=True, fft=True, alpha=0.05, bartlett_confint=True, missing='none')
  plt.figure(figsize=(10,5))
  plot_acf(
      squared_values,
      lags=18,
      alpha=0.05
    )
  plt.show()

```

```python
def hist_risk(percent_vec,alpha):
  if len(percent_vec) <100:
    print(f"lack of data may lead to insufficient statistical reliability")
  w = np.sort(percent_vec)
  q = int(np.floor(alpha*len(percent_vec)))
  VaR = abs(w[q])

#move onto expected shortfall completing our historical risk fucntion
  E_SF = abs(np.mean(w[:q]))
  print(f"The historical {alpha*100}% VaR for {len(percent_vec)} days is: {VaR}")
  print(f"The historical {alpha*100}% Expected shortfall for {len(percent_vec)} days is: {E_SF}")
  return VaR, E_SF
```

```python
def student_risk(value, alpha, weight, percent, kurtosis):
  DF = (6/kurtosis)+4
  sd =  np.sqrt(weight.T@np.cov(percent, rowvar=True)@ weight)
  VaR = -sd*t.ppf(alpha,DF)*value

  #ES
  ES = sd*(t.pdf(t.ppf(alpha,DF),DF)/alpha)*((DF+ t.ppf(alpha,DF)**2)/(DF-1))*value
  print(f"The t dist {alpha*100}% VaR for {percent.shape[1]} days is: {VaR}")
  print(f"The t dist {alpha*100}% Expected shortfall for {percent.shape[1]} days is: {ES}")
  return VaR, ES
```

```python

def EWMA(percent, weight, value, alpha):
  T = percent.shape[1]
  l = 0.94
  cov =  np.cov(percent, rowvar=True)
  for t in range(1,T):
    returns_at_t_minus_1 = percent[:, t-1].reshape(-1, 1) # Make it a column vector
    cov = l * cov + (1 - l) * (returns_at_t_minus_1 @ returns_at_t_minus_1.T)

  sigma = np.sqrt(weight.T@cov@weight)
  VaR = -sigma*norm.ppf(alpha)*value
  return VaR
```

```python
def GARCH(percent_vec, weight, value, alpha):
  # Ensure percent_vec is a pandas Series for arch_model, if it's a numpy array, convert it.
  if not isinstance(percent_vec, pd.Series):
    percent_vec_series = pd.Series(percent_vec)
  else:
    percent_vec_series = percent_vec

  GARCH_model = arch_model(percent_vec_series,mean = 'constant', lags=0, vol = 'GARCH', p=1, q=1, o=0, power= 2.0, dist = 'normal', rescale = True)
  res = GARCH_model.fit(update_freq=5, disp="off")
  print(res.summary())

# 1. Extract the fitted parameters
  omega = res.params['omega']
  garch_alpha = res.params['alpha[1]'] # Renamed to avoid conflict with VaR alpha
  garch_beta = res.params['beta[1]']   # Renamed to avoid conflict with VaR alpha

# 2. Get the "Shock" (y_t)
# This is the return from the very last day in your dataset
  y_t = percent_vec_series.iloc[-1] # Use iloc for the Series

# 3. Get the "Memory" (h_t)
# The library stores the historical standard deviations.
# We grab the very last one and square it to get the variance.
  h_t = res.conditional_volatility.iloc[-1] ** 2

# 4. Compute tomorrow's variance (sigma^2_t+1)
# This is the exact mathematical formula from the textbook
  sigma2_forecast = omega + (garch_alpha * (y_t ** 2)) + (garch_beta * h_t)

# 5. Convert back to Standard Deviation for your VaR formula
  sigma_forecast = np.sqrt(sigma2_forecast)
  var = -sigma_forecast*norm.ppf(alpha)*value # Use the original 'alpha' argument for VaR
  return var
```
